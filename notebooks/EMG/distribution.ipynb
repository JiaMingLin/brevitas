```python
# draw the distribution of the data
import os
import matplotlib.pyplot as plt
import numpy as np

# Load your data, concate g_id_0, g_id_1, g_id_2, g_id_3, g_id_4, along dim=0
train_data = np.load(os.path.join('processed', 'g_id_0.npy'))[:30, :, :]  # Replace with your actual data loading method
train_data = np.concatenate((train_data, np.load(os.path.join('processed', 'g_id_1.npy'))[:30, :, :]), axis=0)
train_data = np.concatenate((train_data, np.load(os.path.join('processed', 'g_id_2.npy'))[:30, :, :]), axis=0)
train_data = np.concatenate((train_data, np.load(os.path.join('processed', 'g_id_3.npy'))[:30, :, :]), axis=0)
train_data = np.concatenate((train_data, np.load(os.path.join('processed', 'g_id_4.npy'))[:30, :, :]), axis=0)
train_data = np.concatenate((train_data, np.load(os.path.join('processed', 'g_id_5.npy'))[:30, :, :]), axis=0)
train_data = np.concatenate((train_data, np.load(os.path.join('processed', 'g_id_6.npy'))[:30, :, :]), axis=0)
train_data = np.concatenate((train_data, np.load(os.path.join('processed', 'g_id_7.npy'))[:30, :, :]), axis=0)


# Plot the distribution of the data
print(train_data.shape)
plt.hist(train_data.flatten(), bins=100, edgecolor='black')

```

```python
# also load the test data
test_data = np.load(os.path.join('processed', 'g_id_0.npy'))[30:, :, :]  # Replace with your actual data loading method
test_data = np.concatenate((test_data, np.load(os.path.join('processed', 'g_id_1.npy'))[30:, :, :]), axis=0)
test_data = np.concatenate((test_data, np.load(os.path.join('processed', 'g_id_2.npy'))[30:, :, :]), axis=0)
test_data = np.concatenate((test_data, np.load(os.path.join('processed', 'g_id_3.npy'))[30:, :, :]), axis=0)
test_data = np.concatenate((test_data, np.load(os.path.join('processed', 'g_id_4.npy'))[30:, :, :]), axis=0)
test_data = np.concatenate((test_data, np.load(os.path.join('processed', 'g_id_5.npy'))[30:, :, :]), axis=0)
test_data = np.concatenate((test_data, np.load(os.path.join('processed', 'g_id_6.npy'))[30:, :, :]), axis=0)
test_data = np.concatenate((test_data, np.load(os.path.join('processed', 'g_id_7.npy'))[30:, :, :]), axis=0)

# Plot the distribution of the data
print(test_data.shape)
plt.hist(test_data.flatten(), bins=100, edgecolor='black')
```

```python
# subplot distribution of each gesture
plt.figure(figsize=(12, 6))
for i in range(8):
    plt.subplot(2, 4, i + 1)
    plt.hist(train_data[i].flatten(), bins=100, edgecolor='black')
    plt.title(f'Gesture {i}')
plt.show()
```
